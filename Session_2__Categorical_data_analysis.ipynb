**Frequency tabel**

```python
con=Visa_df["continent"]=="Asia"
len(Visa_df[con])
```

```python
labels=Visa_df["continent"].unique()
count=[]
for i in labels:
    con=Visa_df["continent"]==i
    count.append(len(Visa_df[con]))
count
```

```python
continent_df=pd.DataFrame(zip(labels,count),columns=["Continent","No of applicants"])
continent_df.to_csv("continent_df.csv",index=False)
```

**groupby**

```python
# dividing continent columns in group
# then calculating size of each group
Visa_df.groupby("continent").size()
```

```python
Continent_keys=Visa_df["continent"].value_counts().keys()
Continent_values=Visa_df["continent"].value_counts().values
Continent_df=pd.DataFrame(zip(Continent_keys,Continent_values),columns=["Continent","No of applicants"])
Continent_df.to_csv("Continent.csv",index=False)
```

**Bar chart under matplotlib**

- plt.bar

```python
Continent_keys=Visa_df["continent"].value_counts().keys()
Continent_values=Visa_df["continent"].value_counts().values
colurs=["green","blue","yellow","orange","red","violet"]
plt.figure(figsize=(10,5))
plt.bar(Continent_keys,Continent_values,color=colurs)
plt.xlabel("Continents")
plt.ylabel("No Of Applicants")
plt.title("Continent_bar")
plt.savefig("Continent_bar.jpg")
plt.show()
```
