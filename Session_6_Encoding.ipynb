```python
path="Visadataset.csv"
visa_df=pd.read_csv(path)
visa_df["continent"].unique()

d={'Africa':0,'Asia':1,'Europe':2,'North America':3,'Oceania':4,'South America':5}
visa_df["continent"]=visa_df["continent"].map(d)
visa_df

# but here i created d maunally automate the code
```

```python
path="Visadataset.csv"
visa_df=pd.read_csv(path)
labels=sorted(visa_df["continent"].unique())
d={}
for i in range(len(labels)):
    keys=labels[i]
    values=i
    d[keys]=values
d
visa_df["continent"]=visa_df["continent"].map(d)
visa_df
```

**Same method with- Dictionary comprehension**

```python
path="Visadataset.csv"
visa_df=pd.read_csv(path)
labels=sorted(visa_df["continent"].unique())
d={labels[i]:i for i in range(len(labels))}
d
visa_df["continent"]=visa_df["continent"].map(d)
visa_df
```

```python
path="Visadataset.csv"
visa_df=pd.read_csv(path)
for i in cat:
    labels=sorted(visa_df[i].unique())
    d={labels[i]:i for i in range(len(labels))}
    d
    visa_df[i]=visa_df[i].map(d)
visa_df
```
