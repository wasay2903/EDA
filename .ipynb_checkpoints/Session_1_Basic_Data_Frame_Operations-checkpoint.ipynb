**Step-2:Create a Data Frame using list**

- Data : your data can be either list or dict

- index : we need to provide a list 

- Columns : we need to provide a list

```python
names=["Ramesh","Suresh","Satish"]
pd.DataFrame(names,columns=["Names"])
```

**Step-3:change the index**

```python
pd.DataFrame(names,index=["A","B","C"],columns=["Names"])
```

```python
idx=["A","B","C"]
cols=["Names"]
pd.DataFrame(names,idx,cols)
```

```python
idx=[1,2,3]
cols=["Names"]
pd.DataFrame(names,idx,cols)
```

**Step-4:Add Multiple columns**

```python
names=["Ramesh","Suresh","Sathish"]
age=[20,22,24]
pd.DataFrame(zip(names,age),
            columns=["Names","Age"])
```

**Step-5:Create empty dataframe and update the columns**

```python
df=pd.DataFrame()
df["Names"]=["Ramesh","Suresh","Sathish"]
df["Age"]=[20,22,24]
df
```

**Step-6:Add a new column with exsisting data frame**

```python
names=["Ramesh","Suresh","Sathish"]
age=[20,22,24]
pd.DataFrame(zip(names,age),columns=["Names","Age"])
city=["mub","hyd","chenn"]
pd.DataFrame(zip(names,age,city),columns=["Names","Age","City"])
```

```python
names=["Ramesh","Suresh","Sathish"]
age=[20,22,24]
df=pd.DataFrame(zip(names,age),columns=["Names","Age"])
df["City"]=["mum","hyd","chen"]
df
```
