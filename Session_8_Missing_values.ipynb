**Fill with methods**

- when ever we are using methods no need to provide values 

$Methods Names$ 

- pad

- bfill

- ffill

- backfill

```python
import warnings
warnings.filterwarnings("ignore")
df.fillna(method="pad")
# it is filling the values of specific coloumn with its is previous value
```

```python
import warnings
warnings.filterwarnings("ignore")
df.fillna(method="pad",axis=1)
# it is filling the values of specific coloumn with its is previous value
# axis 1 means rows wise fill karta
# by default axis 0 columns wise fill karta
```

```python
import warnings
warnings.filterwarnings("ignore")
df.fillna(method="bfill")
# it is fiiling the value based on the preceding value but in reverse direction
```

```python
import warnings
warnings.filterwarnings("ignore")
df.fillna(method="ffill")
# it is filling the values of each column based on the preceding value but in forward direction
# it is same as pad
```

```python
import warnings
warnings.filterwarnings("ignore")
df.fillna(method="backfill")
# it is same as bfill
```

**Fill with mean median mode**

- instead of filling with random values 

- we can use mean median and mode also 

- For numerical columns we mean and median

    - if Data has outlier then use median

- For categorical columns use mode

```python
age_mean=df["Age"].mean()
df["Age"].fillna(age_mean)
```

```python
age_median=df["Age"].median()
df["Age"].fillna(age_median)
```

**KNN Imputere**

- K -Nearest Neighbours

- where k = hyper parameter which means the user can change

- idea :Instead of taking all the samples average 

    - First Fix the neighbour number ie k 

    - Then find the nieghbours using the distance metric 

    - Then take the average of those samepls to fill the missing values

```python
import sklearn
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import KNNImputer
```

```python
import numpy as np
import pandas as pd
from sklearn.impute import KNNImputer
X = [[1,2,np.nan],[3,np.nan,3],[np.nan,60,5],[8,8,7]]
pd.DataFrame(X)
```

```python
imputer=KNNImputer(n_neighbors=2,weights="uniform")
imputer.fit_transform(X)
```

```python
imputer=KNNImputer(n_neighbors=2,weights="distance")
imputer.fit_transform(X)
```
