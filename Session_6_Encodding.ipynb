**Label Encoder Vs OneHotEncoder**

- generally we dont want a relation between input column

- we need both input columns are not correlated to each other 

- we need both input columns are **independent(X,Y)** each other

- we need both input columns are **perpendicular** each other

- we need both input columns are **orthogonal** each other

- we need both input columns are **90 degree phase shift** each other

- OneHotEncoder will achive above

- But OneHotEncoder creates new columns

    - columns are increases means dimenssions are increases

    - More dimenssion data need more processing time

    - More dimenssion data need more infra

    - This is called Curse of dimensionality
