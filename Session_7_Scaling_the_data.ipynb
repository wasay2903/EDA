```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path='Visadataset.csv'
visa_df=pd.read_csv(path)
cat=visa_df.select_dtypes(include='object').columns
num=visa_df.select_dtypes(exclude='object').columns
```

- we have two types of scalling 

    - Z score (Standardization)

    - Min Max Scalar (Normalization)

```python
z = (x-mean/standard deviation)
# x is score
```

```python
pwage=visa_df["prevailing_wage"]
mean=np.mean(pwage)
standard_deviation=np.std(pwage)
d1=pwage-mean
d2=d1/standard_deviation
visa_df["prevailing_wage_z"]=d2
visa_df
```

```python
visa_df[["prevailing_wage","prevailing_wage_z"]]
```

```python
plt.figure(figsize=(20,8))
plt.subplot(2,2,1).hist(visa_df["prevailing_wage"])
plt.subplot(2,2,2).boxplot(visa_df["prevailing_wage"],vert=False)
plt.subplot(2,2,3).hist(visa_df["prevailing_wage_z"])
plt.subplot(2,2,4).boxplot(visa_df["prevailing_wage_z"],vert=False)
plt.show()
```

**sklearn**

```python
from sklearn.preprocessing import StandardScaler
S_s=StandardScaler()
visa_df["prevailing_wage_ss"]=S_s.fit_transform(visa_df[["prevailing_wage"]])
```

```python
visa_df
```

```python
visa_df[["prevailing_wage","prevailing_wage_z","prevailing_wage_ss"]]
```

**Normalization**

```python
wage_data=visa_df["prevailing_wage"]
wage_min=wage_data.min()
wage_max=wage_data.max()
d1=(wage_data-wage_min)
d2=(wage_max-wage_min)
norm=d1/d2
visa_df["prevailing_wage_norm"]=norm
visa_df
```

```python
from sklearn.preprocessing import MinMaxScaler
minmax=MinMaxScaler()
visa_df["prevailing_wage_Norm"]=minmax.fit_transform(visa_df[["prevailing_wage"]])
visa_df
```
