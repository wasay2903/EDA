```python
for i in cat_cols[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    df=pd.DataFrame(zip(keys,values),columns=["label","No of applicants"])
    
    
    folder_name="New_Frequency_table"
    os.makedirs(folder_name,exist_ok=True)
    
    cwd=os.getcwd()
    folder_name="New_Frequency_table"
    new_dir=os.path.join(cwd,folder_name)
    file_name=f"{i}.csv"
    full_path=os.path.join(new_dir,file_name)
    
    df.to_csv(full_path,index=False)
```

**bar chart auto mate**

```python
for i in cat_cols[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    plt.figure(figsize=(10,4))
    plt.bar(keys,values)
    plt.xlabel("labels")
    plt.ylabel("count")
    plt.title(f"{i}_barchart")
    plt.savefig(f"{i}_bar.jpg")
```
